### 🚀 Step 2: Run Live Cloud Extraction (`unitbytes/1688-scraper`)

```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/1688-scraper?fpr=939u3w&fp_sid=gh_1688_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "queries": ["phone case"],
    "maxItems": 30,
    "detailLevel": "full",
    "sortBy": "normal",
    "verticalProductFlag": "factory"
}

print("Running unitbytes/1688-scraper...")
run = client.actor("unitbytes/1688-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("1688-scraper-results.csv", index=False)
df_live.head(10)
```
